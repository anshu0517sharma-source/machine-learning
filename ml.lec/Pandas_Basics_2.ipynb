## 3️⃣ `loc` — selecting by label

> **Flow:** row by label → rows and columns by name → **the slicing difference** ⭐

````python
# loc — one row by its label
df.loc[5:10,'total_bill':'time']
````

**What just happened**

```python
df.loc[5:10, 'total_bill':'time']
```

Two things at once here:

1. `5:10` are **labels**, not positions — and `loc` **includes** the end. That is **6** rows, not 5.
2. `'total_bill':'time'` slices the **columns by name**, giving you every column from
   `total_bill` through `time` in the order they sit in the table.

⭐ This is the thing `iloc` cannot do. `iloc` would need `df.iloc[5:11, 0:6]` — and you would
have to count the columns yourself.

````python
# loc — rows and columns by name
````

### ✋ Guess first, then run

`df.iloc[2:5]` and `df.loc[2:5]`



````python
print(df.iloc[2:5])
print(df.loc[2:5])
````

**The answer: different.**

- `df.iloc[2:5]` → **3** rows (2, 3, 4). It stops *before* 5.
- `df.loc[2:5]` → **4** rows (2, 3, 4, 5). It *includes* 5.

⭐ `loc` includes the end because labels are names, and it makes no sense to ask for
"everything up to but not including the row called 5". `iloc` follows normal Python slicing.

Remember it as: **loc = label = includes. iloc = index = excludes.**

## 4️⃣ Boolean Indexing

So far you knew **which** row you wanted. Now you don't — you only know **what must be true** about it.

> **Flow:** the True/False list → use it to filter → count → two conditions with `&` and `|` → flip it with `~`

````python
df.head(2)
````

````python
# the condition on its own — True/False for every row
df[(df['day']=='Sun') & (df['size']>3)].shape[0]

df[(df['day']=='Sun') | (df['time']=='Dinner')].shape[0]
````

**What just happened**

A condition on its own does not filter anything — it produces a **True/False value for every row**:

```python
df['day'] == 'Sun'      # 244 True/False values
```

Put that inside `df[...]` and Pandas keeps only the True ones. Put `.sum()` on it and you get
a count, because True counts as 1 and False as 0.

Two conditions:

```python
df[(df['day']=='Sun') & (df['size']>3)]     # &  = both must be true
df[(df['day']=='Sun') | (df['time']=='Dinner')]  # |  = either one
```

⭐ **Brackets around each condition are not optional.** `&` is evaluated before `>` in Python,
so `df['size'] > 2 & df['tip'] > 4` tries to do `2 & df['tip']` and dies.

````python
# use it to filter
````

````python
# count how many matched
````

````python
# two conditions — & and |
````

#### Flipping a condition with `~`

You can find the Sunday bills. Now find every bill that was **not** on a Sunday — without using `!=`.

`~` reads as **"not"**. Every True becomes False, every False becomes True.

````python
# ~ flips the whole condition
df[~(df['day']=="Sun")].head()
````

`~` means **not**. It flips every True to False and every False to True.

`df[~(df['day']=='Sun')]` → **168** rows, because 76 of the 244 bills were on Sunday.

Same result as `df[df['day'] != 'Sun']`, but `~` works on any condition, however long —
you wrap the whole thing once instead of rewriting each comparison.

## 📌 Recap — everything from today on one screen

### Picking columns

| You write | You get |
|---|---|
| `df['tip']` | a **Series** — one column |
| `df[['tip']]` | a **DataFrame** — a table with one column |
| `df[['day','tip']]` | a table with two columns |
| `df['new'] = ...` | creates a new column |

### `loc` vs `iloc` ⭐

| | `loc` | `iloc` |
|---|---|---|
| Picks by | **label** (name) | **position** (number) |
| Slice end | **included** | **not included** |
| Columns given as | names | numbers |
| Takes a condition | **yes** | no |

`df.loc[10:20]` gives **11** rows. `df.iloc[10:20]` gives **10**. This is the single most
common mistake in Pandas.

`loc` can also slice **columns** by name: `df.loc[5:10, 'total_bill':'time']`.

### Boolean indexing

Write a condition, get True/False for every row, keep the True ones.

```python
df[df['tip'] > 6]              # filter
(df['tip'] > 6).sum()          # count — True counts as 1
df[(df['size'] > 2) & (df['tip'] > 4)]     # both true
df[(df['time'] == 'Dinner') | (df['size'] > 4)]   # either true
df[~(df['day'] == 'Sun')]      # NOT
```

⭐ Every condition needs **its own brackets**. `&` runs before `>`, so leaving them out breaks it.

### `loc` with a condition — the most used line in Pandas

```python
df.loc[df['tip'] > 6, ['day','tip']]        # rows by condition, columns by name
df.loc[df['day'] == 'Sun', 'total_bill'].mean()
```

### Counting

```python
df['day'].value_counts()                          # count each value
df[df['time'] == 'Dinner']['day'].value_counts()  # filter first, then count
```

### Changing values

```python
df['free_dessert'] = 'No'                          # default for everyone
df.loc[df['size'] >= 6, 'free_dessert'] = 'Yes'    # overwrite where true
df.loc[5, 'tip'] = 5.00                            # fix one single cell
```
