```python
def validate_username(username):
    username = username.strip()

    if len(username) >= 5 and username.isalnum():
        return username.lower()
    else:
        return "Invalid"


print(validate_username(" CodeMaster99 "))

print(validate_username(" Python_Dev "))
```

```python
def reverse_spy_code(secret_code):
    words = secret_code.split("-")
    words.reverse()

    for i in range(len(words)):
        words[i] = words[i].upper()

    return " ".join(words)


print(reverse_spy_code("charlie-bravo-alpha"))
```

```python
def clean_names(raw_names):
    names = raw_names.split(";")
    cleaned_names = []

    for name in names:
        name = name.strip()
        name = name.title()
        cleaned_names.append(name)

    return ", ".join(cleaned_names)

print(clean_names(" john doe ; jAnE sMiTh; ALAN tUrInG "))
```

```python
def remove_extra_spaces(text):
    words = text.split()
    sentence = " ".join(words)
    return sentence.capitalize()

print(remove_extra_spaces(" python is a very powerful language "))
```

```python
def check_palindrome(sentence):
    cleaned = ""

    for character in sentence:
        if character.isalnum():
            cleaned += character.lower()

    return cleaned == cleaned[::-1]

print(check_palindrome("Was it a car or a cat I saw?"))

print(check_palindrome("Python is fun"))

```

```python
def extract_hashtags(post):
    words = post.split()
    hashtags = []

    for word in words:
        if word.startswith("#"):
            hashtag = word.replace("#", "")
            hashtag = hashtag.lower()
            hashtags.append(hashtag)

    return hashtags
    
print(extract_hashtags("Loving the SUNSET! #Nature #Sunset_Vibes #beautiful"))
```

```python
def find_extensions(filenames):
    files = filenames.split(",")
    extensions = []

    for file in files:
        file = file.strip()
        parts = file.split(".")

        extension = parts[-1].upper()

        if extension not in extensions:
            extensions.append(extension)

    return extensions
    
print(find_extensions("report.pdf, image.JPG, data.csv, notes.txt, photo.jpg"))

```

```python
def get_domain(email):
    if "@" not in email:
        return "Invalid"

    parts = email.split("@")
    domain_part = parts[1]

    domain = domain_part.split(".")[0]

    return domain.title()

print(get_domain("developer_99@microsoft.com"))

print(get_domain("admin_at_website.org"))

```

```python
def check_anagram(str1, str2):
    str1 = str1.replace(" ", "").lower()
    str2 = str2.replace(" ", "").lower()

    return sorted(str1) == sorted(str2)

print(check_anagram("Dormitory", "Dirty room"))
```

```python
def format_phone(raw_num):
    cleaned = raw_num

    cleaned = cleaned.replace(" ", "")
    cleaned = cleaned.replace("(", "")
    cleaned = cleaned.replace(")", "")
    cleaned = cleaned.replace(".", "")
    cleaned = cleaned.replace("-", "")
    cleaned = cleaned.replace("+", "")

    if len(cleaned) == 10 and cleaned.isdigit():
        return "(" + cleaned[0:3] + ") " + cleaned[3:6] + "-" + cleaned[6:10]
    else:
        return "Invalid Number"

print(format_phone(" +1 (555) 123.4567 "))

print(format_phone("555-987-6543"))
```

```python
def inspect_sku(sku):
    sku = sku.strip()
    sku = sku.upper()

    if len(sku) == 8:
        if (sku[0:3].isalpha() and
                sku[3] == "-" and
                sku[4:8].isdigit()):

            return "Valid: " + sku

    return "Invalid SKU"

print(inspect_sku(" xYz-9876 "))

print(inspect_sku("AB-12345"))

print(inspect_sku("A@C-1234"))
```
