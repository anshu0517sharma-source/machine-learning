```python
city = " nEw dELhi "

city = city.strip().title()

valid = True

for ch in city:
    if not (ch.isalpha() or ch == " "):
        valid = False
        break

if valid:
    print(city)
else:
    print("Invalid city")
```

```python
prices = " 120, 45 , 300 ,80, 95 "

price_list = []

for item in prices.split(","):
    price_list.append(int(item.strip()))

total = sum(price_list)
highest = max(price_list)
average = total / len(price_list)

print("Total :", total)
print("Highest :", highest)
print(f"Average : {average:.2f}")
```

```python
username = " Rahul123 "

username = username.strip().lower()

if (
    username.isalnum()
    and username[0].isalpha()
    and 5 <= len(username) <= 12
):
    print("Valid username")
else:
    print("Invalid username")
```

```python
sentence = "Python is a very powerful programming language"

words = sentence.split()

count = len(words)

vowels = "aeiouAEIOU"

ending_vowel = 0

for word in words:
    if word[-1] in vowels:
        ending_vowel += 1

longest = words[0]

for word in words:
    if len(word) > len(longest):
        longest = word

print("Total words :", count)
print("Words ending with vowel :", ending_vowel)
print("Longest word :", longest)
```

```python
text = "Aba c"

text = text.lower()

freq = {}

for ch in text:
    if ch != " ":
        if ch in freq:
            freq[ch] += 1
        else:
            freq[ch] = 1

print(freq)
```

```python
sentence = "python is easy python is powerful python"

words = sentence.lower().split()

freq = {}

for word in words:
    if word in freq:
        freq[word] += 1
    else:
        freq[word] = 1

print("Word Frequency:")
print(freq)

most = sorted(freq.items(), key=lambda x: x[1], reverse=True)

print("Most frequent word :", most[0][0])
```

```python
sales = {
    "mon": 200,
    "tue": 350,
    "wed": 200
}

result = {}

for day, amount in sales.items():
    if amount in result:
        result[amount].append(day)
    else:
        result[amount] = [day]

print(result)
```

```python
test1 = {
    "amit": 40,
    "bina": 35
}

test2 = {
    "amit": 30,
    "chirag": 45
}

result = test1.copy()

for student, marks in test2.items():
    if student in result:
        result[student] += marks
    else:
        result[student] = marks

print(result)
```

```python
employees = [
    ("Ravi", 55000),
    ("Sita", 72000),
    ("Amit", 48000)
]

employees = sorted(employees, key=lambda x: x[1], reverse=True)

for name, salary in employees:
    print(f"{name} earns Rs. {salary}")
```

```python
movies = [
    {"title": "Movie A", "rating": 8.2, "genre": "Thriller"},
    {"title": "Movie B", "rating": 8.4, "genre": "Sports"},
    {"title": "Movie C", "rating": 8.1, "genre": "Comedy"},
    {"title": "Movie D", "rating": 8.2, "genre": "Drama"},
]

print("Movies with rating > 8.1")

for movie in movies:
    if movie["rating"] > 8.1:
        print(movie["title"])

genres = []

for movie in movies:
    if movie["genre"] not in genres:
        genres.append(movie["genre"])

print("\nGenres:")
print(genres)

group = {}

for movie in movies:
    rating = movie["rating"]

    if rating in group:
        group[rating].append(movie["title"])
    else:
        group[rating] = [movie["title"]]

print("\nMovies Grouped by Rating")

for rating in sorted(group):
    print(f"Rating {rating} : {group[rating]}")
```
