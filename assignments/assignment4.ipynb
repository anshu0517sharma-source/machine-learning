```python
import numpy as np

marks = np.array([[88, 96, 95],
                  [72, 99, 91],
                  [98, 85, 97],
                  [60, 75, 80]])

bonus = np.array([5, 2, 8])

updated_marks = marks + bonus

capped_count = np.sum(updated_marks > 100)

updated_marks = np.clip(updated_marks, None, 100)

print("Updated Marks:")
print(updated_marks)

print("\nShape:", updated_marks.shape)
print("Marks Capped:", capped_count)
```

```python
import numpy as np

X = np.array([[50, 1200, 3],
              [80, 1500, 9],
              [65, 900, 5],
              [95, 2000, 7]])

col_min = X.min(axis=0)
col_max = X.max(axis=0)

normalized = (X - col_min) / (col_max - col_min)

print("Normalized Matrix:")
print(normalized)

print("\nShape:", normalized.shape)

print("\nColumn Minimum:")
print(normalized.min(axis=0))

print("\nColumn Maximum:")
print(normalized.max(axis=0))
```

```python
import numpy as np

scores = np.array([94, 61, 77, 45, 90, 88, 59, 72])

medals = np.where(scores >= 90, "Gold",
          np.where(scores >= 75, "Silver",
          np.where(scores >= 60, "Bronze", "None")))

print("Medals:")
print(medals)

print("\nGold :", np.sum(medals == "Gold"))
print("Silver :", np.sum(medals == "Silver"))
print("Bronze :", np.sum(medals == "Bronze"))
print("None :", np.sum(medals == "None"))
```

```python
import numpy as np

A = np.array([[1,2,3],
              [4,5,6]])

B = np.array([[7,8,9],
              [1,0,2]])

result1 = A @ B.T

result2 = A.T @ B

print("A @ B.T")
print(result1)
print("Shape:", result1.shape)

print("\nA.T @ B")
print(result2)
print("Shape:", result2.shape)

# A @ B is not possible because:
# A is (2,3)
# B is (2,3)
# Inner dimensions (3 and 2) do not match.
```

```python
import numpy as np

hours = np.array([2,8,5,10,3,7])

attendance = np.array([55,92,70,98,60,85])

hours_col = hours.reshape(-1,1)
attendance_col = attendance.reshape(-1,1)

print("Hours Shape:", hours_col.shape)
print("Attendance Shape:", attendance_col.shape)

table = np.hstack((hours_col, attendance_col))

print("\nTwo Column Table:")
print(table)
print("Shape:", table.shape)

total = table.sum(axis=1).reshape(-1,1)

print("\nTotal Column:")
print(total)
print("Shape:", total.shape)

final_table = np.hstack((table, total))

print("\nThree Column Table:")
print(final_table)
print("Shape:", final_table.shape)

w = np.array([0.5,0.2,0.3])

scores = final_table @ w

print("\nScores:")
print(scores)
print("Shape:", scores.shape)

average = scores.mean()

result = np.where(scores > average, 1, 0)

print("\nAverage Score:", average)

print("Above Average (1=True, 0=False)")
print(result)
```

```python
import numpy as np

m = np.array([[4, 8, 15],
              [16, 23, 42]])

flat = m.flatten()

rav = m.ravel()

flat[0] = 999
rav[0] = 555

print("Flatten Array:")
print(flat)

print("\nRavel Array:")
print(rav)

print("\nOriginal Matrix:")
print(m)

# flatten() creates a copy.
# ravel() creates a view whenever possible.
# Therefore, changing ravel() also changes the original matrix.
```

```python
import numpy as np

np.random.seed(7)

temps = np.random.randint(15, 41, 28)

print("Daily Temperatures:")
print(temps)

weeks = temps.reshape(4,7)

print("\nWeekly Temperatures:")
print(weeks)

print("Shape:", weeks.shape)

weekly_avg = weeks.mean(axis=1)

print("\nWeekly Average:")
print(weekly_avg)

print("\nHottest Week Index:", np.argmax(weekly_avg))

print("Highest Average Temperature:", weekly_avg.max())

print("\nDays Above 35°C:", np.sum(temps > 35))
```

```python
import numpy as np

np.random.seed(5)

A = np.random.randint(1,10,(3,3))

B = np.random.randint(1,10,(3,3))

print("Matrix A")
print(A)

print("\nMatrix B")
print(B)

AB = A @ B
BA = B @ A

print("\nA @ B")
print(AB)

print("Shape:", AB.shape)

print("\nB @ A")
print(BA)

print("Shape:", BA.shape)

print("\nAre they equal?")

print(np.array_equal(AB, BA))

# Matrix multiplication is generally NOT commutative.
# That means A @ B is usually not equal to B @ A.
```

```python
import numpy as np

marks = np.array([[45, 28, 67, 80],
                  [33, 90, 55, 41],
                  [72, 66, 30, 58],
                  [88, 79, 91, 84],
                  [25, 40, 38, 52]])

updated = marks.copy()

updated[updated < 35] = 35

print("Updated Marks")
print(updated)

total = updated.sum(axis=1)

print("\nStudent Totals")
print(total)

print("\nTopper Index:", np.argmax(total))

print("Topper Total:", total.max())
```

```python
import numpy as np

arr = np.arange(1,11)

slice_arr = arr[2:6]

copy_arr = arr[2:6].copy()

slice_arr[0] = 999

copy_arr[0] = 999

print("Original Array")
print(arr)

print("\nSlice")
print(slice_arr)

print("\nCopy")
print(copy_arr)

# slice() creates a view.
# copy() creates an independent copy.
# Therefore only the slice changes the original array.
```
