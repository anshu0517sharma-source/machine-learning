```python
# Q7: Count male and female passengers

print("Number of male and female passengers:")
print(t['Sex'].value_counts())
```

```python
# Q8: Compare the mean age with the 50% value (median)

age_mean = t['Age'].mean()
age_median = t['Age'].median()

print("Mean Age:", age_mean)
print("50% (Median) age:", age_median)

if age_mean > age_median:
    print("The mean is larger than the 50% value.")
    print("This suggests that the age distribution is slightly right-skewed.")
elif age_mean < age_median:
    print("The 50% value is larger than the mean.")
else:
    print("The mean and 50% value are equal.")
```

```python
# Q9: Calculate percentage of missing age values
# Total rows = 891
# Non-null age values = 714
# Missing age values = 891 - 714 = 177

total_rows = t.shape[0]
non_missing_age = 714

missing_age = total_rows - non_missing_age
missing_percentage = (missing_age / total_rows) * 100

print("Total rows:", total_rows)
print("Non-missing age values:", non_missing_age)
print("Missing age values:", missing_age)
print("Percentage of age values missing:", missing_percentage)

# Approximately 19.87% of the age values are missing.
```
